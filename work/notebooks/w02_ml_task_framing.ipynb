## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
#A fixed rule could flag content using one simple threshold, such as marking every item with a large negative performance change as declining. However, content performance can depend on multiple signals, including click-through rate, engagement rate, scroll behavior, traffic sources, and search position. A single hand-written threshold would not capture these interactions well. ML can learn patterns across multiple observed signals and provide a consistent probability that an item belongs to the declining class. The output can help the content team prioritize which items to review.
```
